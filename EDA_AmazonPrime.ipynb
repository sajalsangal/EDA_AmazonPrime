````python
#Check DataType of the imdb column to use appropriate statistical model
titles['imdb_score'].dtype
````

````python
#The column data is normally skewed, visualizing this through a histogram
plt.title('Visualization of imdb_score (includes Null)')
sns.histplot(data = titles, x = 'imdb_score', kde = True)
````

````python
#Using mean on groupby 'type' column to fill null values
titles['imdb_score'] = titles['imdb_score'].fillna(titles.groupby('type')['imdb_score'].transform('mean'))
````

````python
titles['imdb_score'].isnull().sum()
````

Alternate Method in testing

````python
titles[titles['imdb_score'].isnull()]
````

````python
for val in titles['type'].unique():
  type_column_values = titles[titles['type'] == val]
  check_skew = type_column_values['imdb_score'].skew()
  #print(check_skew)

  if(check_skew > 0.5 or check_skew < 0.5): #Use median value to fill null
    median = titles[titles['type'] == val]['imdb_score'].median(skipna = True) # median value of the group

    for index, score in titles['imdb_score'].items(): #Iterate through each value in the column
      if pd.isnull(score) and titles.loc[index, 'type'] == val: #If current column value is null and type column matches with val
        titles.loc[index, 'imdb_score'] = median # Fill the null value with median of the group

  else:
    mean = titles[titles['type'] == val]['imdb_score'].mean(skipna = True) # mean value of the group

    for index, score in titles['imdb_score'].items():
      if pd.isnull(score) and titles.loc[index, 'type'] == val:
        titles.loc[index, 'imdb_score'] = mean

````
